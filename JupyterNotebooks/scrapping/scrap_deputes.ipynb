```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
a=pd.read_csv("deputes/deputes_2022.csv",delimiter=';')
```

```python
datas=[]
years=['07','12','17','22']
for i in years:
    str='deputes/deputes_20'+i+'.csv'
    df0=pd.read_csv(str,delimiter=';')[['nom','groupe_sigle','mandat_debut']]
    datas.append(df0)
    
```

```python
df=pd.concat(datas)
good_years=['2007','2012','2017','2022']
df['mandat_debut']=df['mandat_debut'].apply(lambda x:x[:4])
df.dropna(inplace=True)
df=df[df['mandat_debut'].isin(good_years)]
df.drop(columns=['nom'],inplace=True)
df
```
