```python
def partis(value):
    return french_political_parties[value]
```

```python
dep=[]
df=pd.DataFrame()
for i in range(4):
    df0=pd.read_csv(f'../../data/deputes/deputes_{2007+i*5}.csv',delimiter=';')
    df0=df0[['nom','date_naissance','num_deptmt','num_circo','groupe_sigle','place_en_hemicycle','nb_mandats','mandat_debut', 'mandat_fin']]
    
    df=pd.concat([df,df0])
    
```

```python
df.dropna(inplace=True)
```
