```python
accent_mapping = {
    'à': 'a', 'â': 'a', 'ä': 'a', 'å': 'a',
    'ç': 'c',
    'é': 'e', 'è': 'e', 'ê': 'e', 'ë': 'e',
    'î': 'i', 'ï': 'i',
    'ô': 'o', 'ö': 'o',
    'ù': 'u', 'û': 'u', 'ü': 'u',
    'ÿ': 'y'
}
```

```python
def name(value):
    regex=r'\([^)]*\)'
    val=re.sub(regex,'',value).lower()
    reg_acc=r'[àâäåçéèêëîïôöùûüÿ]'
    def substitution(match):
        return accent_mapping[match.group(0)]

    # Remplacer les lettres accentuées
    res = re.sub(reg_acc, substitution, val)
    return res 
```

```python
fin['nom parti']=fin['nom parti'].apply(name)
```
